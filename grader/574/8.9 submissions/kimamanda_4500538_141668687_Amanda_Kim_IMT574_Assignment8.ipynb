**Instructions**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
import seaborn as sns

df = pd.read_fwf('armada.dat',
                 names=('Battle','Year','Portuguese_ships','Dutch_ships','English_ships','ratio','spanish_involved','portugese_outcome'), 
)
df
```

```python
df.isnull().sum()
```

```python
#SVM

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

#assigning features & target
X = df[['Portuguese_ships', 'Dutch_ships', 'English_ships','spanish_involved']]
y = df['portugese_outcome']

#test train split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

#scaling data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
#training svm classifier
svm_classifier = SVC(kernel='linear', C=1.0, random_state=42)
svm_classifier.fit(X_train_scaled, y_train)

#predicting
y_pred = svm_classifier.predict(X_test_scaled)

#print results
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
#Logisitic regression
from sklearn.linear_model import LogisticRegression

#training using a multinomial parameter
log_model = LogisticRegression(max_iter=1000, multi_class='multinomial', solver='lbfgs')
log_model.fit(X_train_scaled, y_train)

#predict
y_pred_log = log_model.predict(X_test_scaled)

#results
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_log))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_log))
```

```python
#decision tree model
from sklearn.tree import DecisionTreeClassifier

tree_model = DecisionTreeClassifier(max_depth=3,random_state=42)

#train and predict
tree_model.fit(X_train, y_train)
y_pred_tree = tree_model.predict(X_test)

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_tree))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_tree))
```

In order to predict Portuguese outcomes in battle based on number of ships and Spanish involvement, I used a SVM model, logistic regression model, and decision tree model. The SVM model had the highest accuracy at 56%, with logistic regression and decision tree both getting 44% accuracy. Though none of these accuracy scores are good, I was surprised that the SVM model had the highest score. However, the decision tree ended up with the highest average F1 score, indicating that the decision tree is more capable of handling the nuances in the dataset compared to the others.
