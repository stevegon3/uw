```python
import pandas as pd
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
```

```python
df = pd.read_csv('LAB_1_tip_data.csv').loc[:, ~pd.read_csv('LAB_1_tip_data.csv').columns.str.contains('^Unnamed')]
```

```python
df
```

```python
X = df[['Ad', 'Joke', 'None']]
y = df['Tip']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=12)
```

```python
dec_tree = DecisionTreeClassifier()
```

```python
lab1_dec_tree = dec_tree.fit(X_train, y_train)
```

```python
print(dec_tree.feature_importances_)
```

```python
y_pred = dec_tree.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
import matplotlib.pyplot as plt
```

```python
plot_tree(lab1_dec_tree, class_names=['No Tip', 'Tip'], filled=True, feature_names=['Ad', 'Joke', 'None'], rounded=True, fontsize=11, proportion=True)


plt.show()
```

### Based on the accuracy score and feature importances it shows that the ad is not important, the joke provides the most information, and no ad or joke doesn't provide a lot of information. The accuracy score shows that the decision tree was only about 72.73% accurate.
