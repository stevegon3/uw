```python
import pandas as pd
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_excel('LAB_4.2_lenses.xlsx')
```

```python
df = df.rename(columns={'X1.1': 'patient_age', 'X1.2':'spectacle_prescrip', 'X1.3': 'astigmatic', 'X1.4': 'tear_prod_rate', 'X3':'types_of_lenses'})
```

```python
print(df.shape)
df.head()
```

```python
X=df[['patient_age', 'spectacle_prescrip', 'astigmatic', 'tear_prod_rate']]
y = df['types_of_lenses']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=12)
```

```python
dec_tree = DecisionTreeClassifier()
```

```python
dec_tree.fit(X_train, y_train)
```

```python
y_pred = dec_tree.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
dec_tree.feature_importances_
```

### The accuracy score based off of the tested model was 80% percent accurate. When I pulled the feature importances it showed that the patient age did not contribute to the type of lenses while spectcle prescription and a stigmatism showed that there were changes to the feature importances that showed both of these atrributes matttered more then predicting the type of lens while the tear product rate had the highest feature importance showing that this feature mattered the most on the decision tree. 

```python
dec_tree_lab2 = dec_tree.fit(X_train, y_train)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
```

```python
plot_tree(dec_tree_lab2, filled=True, feature_names=['Patient Age', 'Spectacle Prescription', 'Astigmatic', 'Tear Production Rate'], 
         class_names=['Hard', 'Soft', 'None'], rounded=True)

plt.show()
```
