```python
import pandas as pd
```

```python
data = ['YELLOW,SMALL,STRETCH,ADULT,T',
'YELLOW,SMALL,STRETCH,ADULT,T',
'YELLOW,SMALL,STRETCH,CHILD,F',
'YELLOW,SMALL,DIP,ADULT,F',
'YELLOW,SMALL,DIP,CHILD,F',
'YELLOW,LARGE,STRETCH,ADULT,T',
'YELLOW,LARGE,STRETCH,ADULT,T',
'YELLOW,LARGE,STRETCH,CHILD,F',
'YELLOW,LARGE,DIP,ADULT,F',
'YELLOW,LARGE,DIP,CHILD,F',
'PURPLE,SMALL,STRETCH,ADULT,T',
'PURPLE,SMALL,STRETCH,ADULT,T',
'PURPLE,SMALL,STRETCH,CHILD,F',
'PURPLE,SMALL,DIP,ADULT,F',
'PURPLE,SMALL,DIP,CHILD,F',
'PURPLE,LARGE,STRETCH,ADULT,T',
'PURPLE,LARGE,STRETCH,ADULT,T',
'PURPLE,LARGE,STRETCH,CHILD,F',
'PURPLE,LARGE,DIP,ADULT,F',
'PURPLE,LARGE,DIP,CHILD,F']
```

```python
split_data = [row.split(',') for row in data]
```

```python
split_data
```

```python
df = pd.DataFrame(split_data, columns=['Color', 'Size', 'Act', 'Age', 'One Class Label-Inflated'])
```

```python
df.head()
```

```python
df.Color.value_counts()
```

```python
df['Color'] = df.Color.replace({'YELLOW':0, 'PURPLE':1})
```

```python
df['Size'] = df.Size.replace({'SMALL':0, 'LARGE':1})
```

```python
X_1 = df[['Color', 'Size']]
y_1 = df['One Class Label-Inflated']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn import metrics
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
```

```python
X1_train, X1_test, y1_train, y1_test = train_test_split(X_1, y_1, test_size=0.25, random_state=12)
```

```python
dec_tree = DecisionTreeClassifier(random_state=12)
```

```python
dectree_1 = dec_tree.fit(X1_train, y1_train)
```

```python
y_pred = dec_tree.predict(X1_test)
```

```python
metrics.accuracy_score(y1_test, y_pred)
```

```python
plot_tree(dectree_1, filled=True, feature_names=['Color', 'Size'], class_names=['Inflated', 'Not_Inflated'], rounded=True)

plt.show()
```

```python
df2 = df.copy()
```

```python
df2['Act'] = df2.Act.replace({'STRETCH':0., 'DIP':1})
```

```python
df2.Age.value_counts()
```

```python
df2['Age'] = df2.Age.replace({'ADULT':1, 'CHILD':0})
```

```python
df2.head()
```

```python
X_2 = df2[['Act', 'Age']]
y_2 = df2['One Class Label-Inflated']
```

```python
X2_train, X2_test, y2_train, y2_test = train_test_split(X_2, y_2, test_size=0.2, random_state=12)
```

```python
dec_tree2 = dec_tree.fit(X2_train, y2_train)
```

```python
y_pred_2 = dec_tree.predict(X2_test)
```

```python
metrics.accuracy_score(y2_test, y_pred_2)
```

```python
plot_tree(dec_tree2, filled=True, feature_names=['Act', 'Age'], class_names=['Inflated', 'Not_Inflated'], rounded=True)

plt.show()
```

#### I plotted both of the trees and they look relatively similar however the acccuracy score for the decision tree that used act and age as the predictors was 100% accurate versus the decision tree that used color and size as its predictors only produced an accuracy score of 20%.
