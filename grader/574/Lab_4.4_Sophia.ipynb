```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('LAB_crash.csv')
```

```python
df_c = df.set_index('PassengerId')
```

```python
df_c.head()
df_c.shape
```

```python
X = df_c[['Speed', 'Age']]
y = df['Survived']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=8)
```

```python
dec_tree = DecisionTreeClassifier(criterion='gini', random_state=12)
```

```python
dec_tree.fit(X_train, y_train)
```

```python
accuracy = dec_tree.score(X_test, y_test)
```

```python
print(accuracy)
```

```python
plt.figure(figsize=(10,6))

plot_tree(dec_tree, feature_names=['Age', 'Speed'], class_names=['Died', 'Survived'])
plt.show()
```

## Test with a lower test_size a greater volume of training data

```python
df_test = df_c.copy()
```

```python
X = df_test[['Speed', 'Age']]
y = df['Survived']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=8)
```

```python
dec_tree = DecisionTreeClassifier(criterion='gini', random_state=12)

dec_tree.fit(X_train, y_train)
```

```python
accuracy = dec_tree.score(X_test, y_test)
print(accuracy)
```

```python
plt.figure(figsize=(10,6))

plot_tree(dec_tree, feature_names=['Age', 'Speed'], class_names=['Died', 'Survived'])
plt.show()
```

### When I first ran the decision tree model to predict survivability of a car crash using age and speed as the predictor I used a test size of 30% of the data and trained using only 70% of the data. This data set is rather small with only 20 rows of the data. This produced an accuracy score of 50%, which was lower than the highest accuracy score that I got from the Logistic Regression and kNN classification models. Both of these models resulted in only a 75% accuracy score. When I adjusted the decision tree training size from 70% of the data to 90% I saw the accuracy score jump from 50% to 100%.
