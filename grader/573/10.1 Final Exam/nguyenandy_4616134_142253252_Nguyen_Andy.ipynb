Instead of modeling the number of affairs, consider the binary outcome - had an affair versus didn't have an affair. Create a new variable to capture this response variable of interest. What might the advantages and disadvantages of this approach to modeling the data be in this context?

```python
#Create a binary outcome variable: 1 = had an affair, 0 = no affair
#We classify anyone with 'affairs' > 0 as having had an affair
affairs['had_affair'] = (affairs['affairs'] > 0).astype(int)
```

To explore who engaged in extramarital affairs, I first created a binary variable that indicates whether a respondent had any affair activity using the 'affairs' variable.
This new variable equals 1 if the respondent reported any amount of affair involvement and 0 otherwise.

Modeling the data using this binary outcome has several advantages. It simplifies interpretation, allowing us to focus on the broader question of whether an affair occurred rather than the exact amount of time spent having affairs, which may be difficult to measure reliably. A binary outcome is also easier to model with logistic regression and tends to be more robust to extreme values, such as unusually high affair scores that could skew results.

However, this approach also has important limitations. By converting a continuous measure into a binary one, we lose information about the intensity or frequency of affairs. Someone who had one brief encounter is treated the same as someone with extensive affair activity, which may mask meaningful differences in behavior. Additionally, the threshold of “greater than zero” assumes that any reported affair, no matter how small, is equally significant, which may not reflect the real-world variation captured in the original data.

(c) Use an appropriate regression model to explore the relationship between having an affair (binary) and other personal characteristics. Comment on which covariates seem to be predictive of having an affair and which do not.

```python
#Import logistic regression tools from statsmodels
import statsmodels.formula.api as smf

#Fit a logistic regression model
#Logistic regression is appropriate when the outcome is binary.
#The model estimates how each predictor changes the log-odds of having an affair.
model = smf.logit(
    formula='''
        had_affair ~ age 
        + yrs_married 
        + children 
        + religious 
        + educ 
        + occupation 
        + occupation_husb 
        + rate_marriage
    ''',
    data=affairs
).fit()

#Print the full model summary table
print(model.summary())
```

From the logistic regression results, several covariates appear to be strong predictors of whether someone had an affair. Years married shows a significant positive coefficient (p < 0.001), indicating that individuals who have been married longer are more likely to report an affair. Religiosity and marital satisfaction (rate_marriage) both have significant negative coefficients (p < 0.001), meaning more religious respondents and those who rate their marriage more positively are less likely to engage in extramarital sex. Age, number of children, and education show small but statistically significant effects, although their practical impact is weaker compared to marital satisfaction and religiosity. In contrast, occupation and husband’s occupation have coefficients with relatively higher p-values and weaker effect sizes, suggesting they are not strong or reliable predictors of having an affair in this dataset. Overall, relationship-oriented variables (years married, religiosity, marital rating) are much more predictive of affair behavior than demographic or socioeconomic variables.

(d) Use a model selection procedure to obtain a "best" fit model. Is the model different from the full model you fit in part (c)? Which variables are included in the "best" fit model? [Hint: stepwise selection] 

```python
#List of all possible predictors
predictors = ["age", "yrs_married", "children", "religious", 
              "educ", "occupation", "occupation_husb", "rate_marriage"]

#Start with an empty model
selected = []

#Stepwise selection using AIC
def compute_aic(vars_list):
    formula = "had_affair ~ " + " + ".join(vars_list)
    model = smf.logit(formula=formula, data=affairs).fit(disp=False)
    return model.aic, model

best_aic = float("inf")
best_model = None

for var in predictors:
    trial_vars = selected + [var]
    aic, model = compute_aic(trial_vars)
    if aic < best_aic:
        best_aic = aic
        best_model = model
        selected.append(var)

print("Selected variables:", selected)
print(best_model.summary())
```

```python
#Get residuals and fitted values
residuals = model_bmi.resid
fitted = model_bmi.fittedvalues

#Residuals vs Fitted plot
plt.figure(figsize=(6,4))
sns.scatterplot(x=fitted, y=residuals)
plt.axhline(0, color='red', linestyle='--')
plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted")
plt.show()

#Histogram of residuals
plt.figure(figsize=(6,4))
sns.histplot(residuals, bins=30, kde=True)
plt.title("Distribution of Residuals")
plt.show()

#QQ plot for normality
sm.qqplot(residuals, line='45')
plt.title("QQ Plot of Residuals")
plt.show()
```

(d) Use a stepwise model selection procedure of your choice to obtain a "best" fit model. Is the model different from the full model you fit in part (b)? If yes, how so?

```python
#Encode categorical variables
df_encoded = pd.get_dummies(df, drop_first=True)

#Stepwise forward selection function
def forward_selection(data, response):
    remaining = list(data.columns)
    remaining.remove(response)
    selected = []
    current_score, best_new_score = 0.0, 0.0
    
    while remaining and current_score == best_new_score:
        scores_with_candidates = []
        for candidate in remaining:
            formula = "{} ~ {}".format(response,
                                       ' + '.join(selected + [candidate]))
            score = sm.OLS.from_formula(formula, data).fit().rsquared_adj
            scores_with_candidates.append((score, candidate))
        
        scores_with_candidates.sort()
        best_new_score, best_candidate = scores_with_candidates.pop()
        
        if best_new_score > current_score:
            remaining.remove(best_candidate)
            selected.append(best_candidate)
            current_score = best_new_score
    
    return selected

best_vars = forward_selection(df_encoded, "charges")
best_vars
```

```python
formula = "charges ~ " + " + ".join(best_vars)
best_model = sm.OLS.from_formula(formula, df_encoded).fit()
best_model.summary()
```

Using forward stepwise selection, the “best” model included several predictors: age, bmi, smoker, children, and some region indicators. This model explains substantially more variance in medical charges than the simple model from part (b).

The stepwise-selected model is very different from the part (b) model, which only included BMI. In the earlier model, BMI explained only about 3.9% of the variance, while the stepwise model includes multiple important predictors—especially smoking status—which is known to have a large effect on medical costs. As a result, the new model provides a much better fit and addresses much more of the variation in the data.

In summary, stepwise selection retained only the strongest predictors and produced a model that is far more informative and accurate than the simple BMI-only model from part (b).

(b) Show that the MLE of $\theta$ does not exists if every observed value is 0.

```python
#all observed values = 0
X = np.zeros(10)

#define the log-likelihood for Poisson(theta)
def loglik(theta, n=len(X)):
    return -n * theta  # because all Xi = 0

#theta values to visualize
theta_vals = np.linspace(0.001, 5, 200)   #start just above 0
ll_vals = loglik(theta_vals)

#plot log-likelihood
plt.plot(theta_vals, ll_vals)
plt.xlabel("theta")
plt.ylabel("log-likelihood")
plt.title("Poisson Log-Likelihood When All Observations = 0")
plt.show()
```

What This Code Shows:

The log-likelihood is a straight decreasing line. The maximum occurs at θ = 0, but the parameter space requires θ > 0. Therefore, there is no valid MLE.
