(c) What are the advantages and disadvantages of a very flexible (versus a less flexible) approach for regression or classification? Under what circumstances might a more flexible approach be preferred to a less flexible approach? When might a less flexible approach be preferred?

##### Answer
A very flexible approach can fit complicated patterns in the data. The main advantage is that it usually gives high prediction accuracy, especially when the relationship between predictors and the response is very nonlinear or hard to capture with simple methods. However, the disadvantage is that these models can be hard to interpret. It’s difficult to explain exactly why the model made a certain prediction.

On the other hand, a less flexible approach is simpler and much easier to interpret. The disadvantage is that they may not capture complex relationships well, so prediction accuracy might be lower if the real-world relationship is nonlinear or complicated.

A more flexible model is preferred when the main goal is prediction, when the underlying pattern is complex, and when we have a large dataset to train on. For example, predicting Instagram engagement or Seattle weather might benefit from flexible models. A less flexible model is preferred when the main goal is interpretation or inference, when we want to understand how variables are related, or when we have limited data. This is useful in situations like understanding which travel factors influence trip cost or analyzing voting behavior.

##### Answer
False.
Even though the GPA–IQ interaction coefficient is only 0.01, that doesn’t automatically mean the interaction isn’t important.
