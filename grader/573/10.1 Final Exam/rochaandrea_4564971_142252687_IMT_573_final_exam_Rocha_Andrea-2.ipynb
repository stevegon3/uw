```python
#removing bmi as a predictor so I can calculate the change in the r-squares to see variance for bmi
prds_no_bmi = ['age', 'children', 'sex_male', 'smoker_yes']
prds_2 = dummy_df[prds_no_bmi]
```

```python
no_bmi_model = LinearRegression()
no_bmi_model.fit(prds_2, rsp)
```

```python
#r-squared of all predictors I decided to include minus the bmi
r_sqrd_2= no_bmi_model.score(prds_2, rsp)
print(r_sqrd_2)
```

```python
#seeing variance for bmi
bmi_variance =  r_sqrd - r_sqrd_2
print("BMI Variance:", bmi_variance)
```

(a) Which answer is correct and why? <font color = red>[single choice question]

i. For a fixed value of IQ and GPA, males earn more on average than females.

(b) Wite down the least square function and then predict the salary of a female with IQ of 110 and a GPA of 4.0.

```python
#function for least squares to predict salary
def salary_prediction(gpa, iq, gender):
    beta_0 = 50
    beta_1 = 20
    beta_2 = 0.07
    beta_3 = 35
    beta_4 = 0.01
    beta_5 = -10

    gpa_iq = gpa * iq
    gpa_gender = gpa * gender

    salary = (beta_0 +
    beta_1 * gpa +
    beta_2 * iq +
    beta_3 * gender +
    beta_4 * gpa_iq +
    beta_5 * gpa_gender)

    return salary
```

```python
#female prediction 
female_prd = salary_prediction(gpa= 4.0, iq= 110, gender= 1)
print(female_prd)
```
