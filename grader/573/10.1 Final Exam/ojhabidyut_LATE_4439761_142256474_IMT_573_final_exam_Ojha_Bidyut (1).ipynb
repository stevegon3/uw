## Model Selection Procedure
# Reference -- https://www.statology.org/stepwise-selection/
# reference-- https://uw.hosted.panopto.com/Panopto/Pages/Viewer.aspx?id=9884004c-b749-4923-bdfc-b217000c37e3&start=1270.1
---https://medium.com/@kelvinsang97/feature-selection-stepwise-regression-forward-selection-and-backward-elimination-with-python-d53230be995c
# We will implement here a Stepwise logistic selection which selects the best subset of predictors using model fit criteria.

```python
# reference (OpenAI, 2025)
def stepwise_selection(X, y, 
                       initial_list=[], 
                       threshold_in=0.01, 
                       threshold_out=0.05, 
                       verbose=True):
    
    included = list(initial_list)
    while True:
        changed=False
        
        # forward step
        excluded = list(set(X.columns) - set(included))
        new_pvals = pd.Series(index=excluded, dtype=float)
        for col in excluded:
            model = sm.Logit(y, sm.add_constant(pd.DataFrame(X[included + [col]]))).fit(disp=0)
            new_pvals[col] = model.pvalues[col]
        best_pval = new_pvals.min() if len(new_pvals)>0 else None
        
        if best_pval is not None and best_pval < threshold_in:
            best_feature = new_pvals.idxmin()
            included.append(best_feature)
            changed=True
            if verbose:
                print("Add  {:20} with p-value {:.4f}".format(best_feature, best_pval))
                
        # backward step
        model = sm.Logit(y, sm.add_constant(pd.DataFrame(X[included]))).fit(disp=0)
        pvals = model.pvalues.iloc[1:]
        worst_pval = pvals.max()
        
        if worst_pval > threshold_out:
            worst_feature = pvals.idxmax()
            included.remove(worst_feature)
            changed=True
            if verbose:
                print("Drop {:20} with p-value {:.4f}".format(worst_feature, worst_pval))
        
        if not changed:
            break
            
    return included
```

```python
X = affairs [['rate_marriage', 'age', 'yrs_married', 'children', 'religious', 'educ', 'occupation', 'occupation_husb']]
y = affairs['yes_affair']

best_vars = stepwise_selection(X, y)
print("\nBest model variables:", best_vars)
```

## The step wise model is different from full model since stepwise best model has only 5 predictors against 8 predictors of full-model.
The variables included in best fit model are rate_marriage,yrs_married, religious, age and occupation.Stepwise selection removed three variables from the full model:
children, educ, and occupation_husb.
Therefore, the best model is smaller and more parsimonious than the full model while still capturing all meaningful predictors of yes_affair.

(d) Use a stepwise model selection procedure of your choice to obtain a "best" fit model. Is the model different from the full model you fit in part (b)? If yes, how so?

# References -- https://canvas.uw.edu/courses/1830727/pages/7-dot-8-variable-selection?module_item_id=24478641
https://github.com/talhahascelik/python_stepwiseSelection/blob/master/stepwiseSelection.py

Stepwise regression iteratively adds or removes predictors based on some criterion (usually AIC, BIC, or p-value) to find a model that balances goodness-of-fit and simplicity.
Akaike information criterion (AIC)
Bayesian information criterion (BIC)
 > Forward selection: Start with no variables, add the most significant predictor at each step.
 > Backward elimination: Start with all variables, remove the least significant predictor at each step.
 > Both (stepwise): Combines forward and backward selection.
>

```python
# Forward selection

# Start with y and full candidate predictors
y = df["charges"]
X = df.drop(columns=["charges"])

# reference -- (OpenAI,2025)
def forward_selection(X, y):
    remaining = list(X.columns)
    selected = []
    current_score, best_new_score = np.inf, np.inf
    
    while remaining:
        scores_with_candidates = []
        
        for candidate in remaining:
            formula_vars = selected + [candidate]
            X_train = sm.add_constant(X[formula_vars])
            model = sm.OLS(y, X_train).fit()
            scores_with_candidates.append((model.aic, candidate))
        
        scores_with_candidates.sort()
        best_new_score, best_candidate = scores_with_candidates[0]

        if best_new_score < current_score:
            remaining.remove(best_candidate)
            selected.append(best_candidate)
            current_score = best_new_score
        else:
            break
            
    return selected
    
# Run Forward selection    
selected_features = forward_selection(X, y)
selected_features
```

```python
# Fit the selected model 
X_selected = sm.add_constant(df[selected_features])
model_fs = sm.OLS(y, X_selected).fit()
print(model_fs.summary())
```

## Forwarded selection model output :R squared = 0.751 and with 6 predictors selected. It doesn't include sex_male, region_northwest predictors since they weren't <p>

1) Can capture complex relationships: Flexible models (e.g., splines, decision trees, neural networks) can fit highly nonlinear patterns between predictors and response.
2) Lower bias: They can adapt closely to the training data, reducing systematic errors.
3) Better performance with large, complex datasets: When there is enough data, flexible models often achieve higher predictive accuracy.

Disadvantages of a very flexible approach
1) Higher variance / overfitting risk: They can fit the noise in the training data, leading to poor generalization.
2) Harder to interpret: Understanding the effect of individual predictors is difficult.
3) Computationally expensive: More parameters often mean longer training times and higher resource usage.

Advantages of a less flexible approach
1) Lower variance / less overfitting: Simple models (e.g., linear regression, logistic regression with few predictors) generalize better on small datasets.
2) Easier to interpret: Coefficients and relationships are transparent, which is useful for inference.
3) Faster and simpler: Less computational burden and simpler to implement.

Disadvantages of a less flexible approach
1) Higher bias: May fail to capture nonlinear relationships, underfitting the data.
2) Poor performance on complex datasets: Simple models can miss subtle patterns, reducing predictive accuracy.

We can opt for more flexible approach when the relationship between predictors and response is complex or nonlinear or when dealing with very large datasets and focussing
