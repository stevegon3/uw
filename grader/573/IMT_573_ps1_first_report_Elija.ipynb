# Problem 1: Load a dataset

Choose a dataset from the sklearn [toy dataset repository](https://scikit-learn.org/stable/datasets/toy_dataset.html). Load the data and provide a brief description of the data. ## The dataset has 178 rows and 14 columms. (this is from Shape: 178, 14) 
## Each row demonstrates one sample of wine. There are 178 samples of wine total

```python
#Loading the dataset from wine
wine = datasets.load_wine()
df = pd.DataFrame(wine.data, columns=wine.feature_names)
df['target'] = wine.target

##This is just printing out the look of the dataset
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
print("First rows:\n", df.head())
```

### Problem 2: Create a visualization

Create a visualization of the data you chose.

```python
import matplotlib.pyplot as plt ##I have to import this since Python doesn't know plt 
```

```python
sns.scatterplot(
    data=df, x="alcohol", y="color_intensity", hue="target",
    palette=sns.light_palette("pink", n_colors=3) ##this is setting the different shades of pink
)
plt.title("Wine types by alcohol and color (By shades of pink)")
plt.show()
```
