### Problem 2: Create a visualization

```python
# Load dataset
diabetes_ds = datasets.load_diabetes(as_frame=True)
# Convert to DataFrame
diabetes = diabetes_ds.frame
# Display first 5 rows
diabetes.head(5)
# Draw histogram of the distribution in the target variable
sns.histplot(diabetes['target']).set(
    xlabel = 'Quantitative Measure of Disease Progression A Year After Baseline',
    ylabel = 'Count',
    title = 'Distribution of The Target Variable'
)
```
