### Problem 2: Create a visualization

Create a visualization of the data you chose.

```python
import matplotlib.pyplot as plt
from sklearn import datasets

iris = datasets.load_iris(as_frame=True)
df = iris.frame

# Scatter: sepal length vs sepal width
plt.figure()
plt.scatter(df[iris.feature_names[0]], df[iris.feature_names[1]], s=20)
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.title("Iris: Sepal length vs. Sepal width")
plt.show()

# Boxplots for all features
plt.figure()
df[iris.feature_names].boxplot(rot=45)
plt.title("Iris: Boxplot of features")
plt.tight_layout()
plt.show()
```
