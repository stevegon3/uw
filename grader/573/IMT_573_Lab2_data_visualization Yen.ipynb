```python
To tackle this problem, we have to ask the following question.
What passenger characteristics or other factors seem to be associated with survival?
For this, we’ll use a grouped bar plot to explore how passenger class (pclass) and gender (sex) relate to survival rates.
This uses two variables other than survival (pclass and sex).
It’s a clear, categorical visualization, different from the violin plot.
```

```python
# import pandas
import pandas as pd

# Set plot style
sns.set_style("whitegrid")

# Compute survival rates grouped by pclass and sex
survival_rates = (
    titanic.groupby(["pclass", "sex"])["survived"]
    .mean()
    .reset_index()
)

# Visualization: Grouped Bar Plot
plt.figure(figsize=(10, 6))

# Bar plot: x = pclass, y = survival rate, hue = sex
sns.barplot(
    x="pclass",
    y="survived",
    hue="sex",
    data=survival_rates,
    palette={"male": "#0000FF", "female": "#00FF00"} # Male Blue and Female Green
)

# Customize axes, legend, and title
plt.title("Survival Rate by Passenger Class and Sex", fontsize=14, pad=15)
plt.xlabel("Passenger Class", fontsize=12)
plt.ylabel("Survival Rate", fontsize=12)
plt.ylim(0, 1)  # Survival is a proportion between 0 and 1

# Set title
plt.legend(title="Sex", loc="upper right")
plt.tight_layout()

# Show the plot
plt.show()
```

**Description of the Visualization**

This grouped bar plot highlights how passenger class and gender jointly influenced survival probabilities aboard the Titanic.

Female passengers had higher survival rates across all classes compared to male passengers. First-class females show the highest survival rate, with more than 90% surviving, while third-class males had the lowest survival rate, well below 20%. The stark contrast between male and female survival within each class reflects the “women and children first” evacuation policy and the overall priority given to first-class passengers.

Additionally, survival rates decline as passenger class decreases for both genders, indicating that Social class played a significant role in survival likelihood. This plot demonstrates that both gender and class were significant determinants of survival on the Titanic.
