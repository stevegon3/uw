### Problem 3: Exploring data

For each of the questions you proposed in Problem 2, perform an exploratory data analysis designed to address the question. At a minimum, you should produce two visualizations related to each question. Be sure to describe what the visuals show and how they speak to your question of interest.

```python
import matplotlib.pyplot as plt

sns.set_style('whitegrid')

#Question 1: What are the different animal types and how many in each category?
animal_counts = df['Animal Type'].value_counts()
plt.figure(figsize=(10, 6))
sns.barplot(x=animal_counts.index, y=animal_counts.values, palette='viridis')
plt.title('Counts of Different Animal Types')
plt.xlabel('Animal Type')
plt.ylabel('Count')
plt.show()  
```

### Q1 Viz
- The visual shows the animal type based on the results and the count. The visualization shows the spread and difference between each animal type and the count in relation to each animal that has been at the shelter.

```python
#Question 2: How many cats were alive at intake? How many cats later died?
cat_intake_counts = cats['outcome_is_dead'].value_counts()
plt.figure(figsize=(8, 5))
sns.barplot(x=cat_intake_counts.index, y=cat_intake_counts.values, palette='magma')
plt.title('How many cats were alive at outcome vs. dead at outcome?')
plt.xlabel('Outcome Status (False = Alive, True = Dead)')
plt.ylabel('Count')
plt.show()
```

### Q2 Viz
- The visual shows the cat count for animal status. The visualization shows the difference between the cats that were alive and that died during outcome. The combination of the two bars give the total of the total cats that were intaken at shelter.
