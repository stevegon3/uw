```python
import seaborn as sns
faithful = sns.load_dataset("geyser").rename(columns={"duration": "eruptions"})
faithful = faithful[["eruptions", "waiting"]]  # keep only the two columns you need
faithful.head()

```

```python
# basic stats
avg_wait = faithful["waiting"].mean()
min_wait = faithful["waiting"].min()
max_eruption = faithful["eruptions"].max()
print(avg_wait, min_wait, max_eruption)

# histogram example
import seaborn as sns, matplotlib.pyplot as plt
ax = sns.histplot(faithful["waiting"], bins=20, kde=True)
ax.set(xlabel="Waiting time (min)", ylabel="Count", title="Distribution of Waiting Time")
plt.show()

```
