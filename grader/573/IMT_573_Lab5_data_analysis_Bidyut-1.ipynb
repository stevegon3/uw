#### (a) Try the easy solution first

After filtering any anomolous values, try using descriptive statistics to see if you can get a general sense of the answer to your question.

```python
# Check columns
data_all.head(5) 

# Add a publisher column
dc_data["Publisher"] = "DC"
marvel_data["Publisher"] = "Marvel"

# Combine datasets
data_all = pd.concat([dc_data, marvel_data])


# Keep relevant columns
data_clean = data_all[["Publisher", "SEX", "Year"]].copy()

# Convert Year to numeric (force errors or NULL values to NaN)
data_clean["Year"] = pd.to_numeric(data_clean["Year"], errors="coerce")


# Drop rows with missing or invalid years
data_clean = data_clean.dropna(subset=["Year"])


# Convert to integer type
data_clean["Year"] = data_clean["Year"].astype(int)


# Filter out years
# data_clean = data_clean[(data_clean["Year"] >= 1930) & (data_clean["Year"] <= 2025)]

# Clean up 'SEX' column to avoid duplicate values in different cases (lower or upper)
data_clean["SEX"] = data_clean["SEX"].astype(str).str.strip().str.title()

#how many male vs. female characters
# print(data_clean["SEX"].value_counts())

"""
SEX
Male Characters           11100
Female Characters          3628
Nan                         795
Agender Characters           36
Genderfluid Characters        2
Name: count, dtype: int64
"""

# Gender counts by publisher
print(data_clean.groupby("Publisher")["SEX"].value_counts())

gender_trend = (
    data_clean.groupby(["Publisher", "Year", "SEX"])
    .size()
    .reset_index(name="Count")
)


print(gender_trend.head())

# Filter for DC publisher, female characters, and valid Year
dc_female = data_all[
    (data_all["Publisher"] == "DC") &
    (data_all["SEX"].str.strip().str.title() == "Female Characters") &
    (data_all["Year"].notnull())
].copy()

# Optional: convert Year to integer if not already done
# dc_female["Year"] = dc_female["Year"].astype(int)

# View the first few rows
print(dc_female.head(20)) # Checking for Dc dataset with above conditions has no data. Hence output has data of Marvel dataset only.
```

#### (b) Check distributions

Look at distribution(s) of your data to determine if there are any patterns that are evident with respect to your question of interest.

```python
# For female Marvel characters since there are none in DC with Year valid column as NULL value.
Marvel_female_per_year = data_clean.groupby("Year").size().reset_index(name="Count")
print(Marvel_female_per_year.head(10))

# Distribution
plt.figure(figsize=(12,5))
sns.lineplot(data=Marvel_female_per_year, x="Year", y="Count", marker="o")
plt.title("Number of Female Marvel Characters Introduced Per Year")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
```

Since relevant data is missing from DC dataset for female characters so we can't do a side by side comparison for both publishers. Based, on this issue, we can pick next Male publisher as an area of interest and apply similar conditions on data and compare. We might face same issue because of missing data from SEX, APPERANCES and YEAR columns. Even though we try to go ahead with a combined dataset to find the results, the problem persists because of presence of two Year columns- which can't be merged into one. 

```python
# Step 1: Filter male characters with valid years
# Ensure Year is numeric
data_all["Year"] = pd.to_numeric(data_all["Year"], errors="coerce")
data_all = data_all.dropna(subset=["Year"])
data_all["Year"] = data_all["Year"].astype(int)

# Normalize SEX column
data_all["SEX"] = data_all["SEX"].astype(str).str.strip().str.title()

# Filter for male characters only
male_characters = data_all[
    (data_all["SEX"] == "Male Characters")].copy()

print(male_characters.head(10))
print(male_characters.shape)
```

```python
# Step 2: Aggregate counts by Publisher, Year, 
male_trend = (
    male_characters.groupby(["Publisher", "Year"])
    .size()
    .reset_index(name="Count")
)
print(male_trend.head(100))


```

```python
#Step3: Visualize 
plt.figure(figsize=(12,6))
sns.lineplot(
    data=male_trend,
    x="Year",
    y="Count",
    hue="Publisher",
    markers="o"
   )
plt.title("Number of Male Characters Introduced Per Decade (DC & Marvel)")
plt.xlabel("Year")
plt.ylabel("Count")
plt.legend(title="Publisher")
plt.show()
```
