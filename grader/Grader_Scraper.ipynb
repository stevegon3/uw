```python
import re
import pandas as pd
from datetime import datetime

text = """
2.7 Discussion: Stand Your Ground (https://canvas.uw.edu/courses/1830727/discussion_topics/9915100) Due Oct 5, 11:59 PM
1.3 Discussion: "Hey, You're Having a Baby!" (https://canvas.uw.edu/courses/1830727/discussion_topics/9915101) Due Sep 28, 11:59 PM
1.1 Discussion: Introductions (https://canvas.uw.edu/courses/1830727/discussion_topics/9915102) Due Sep 28, 11:59 PM
4.7 Data in the Wild (https://canvas.uw.edu/courses/1830727/discussion_topics/9915099) Due Oct 19, 11:59 PM
5.10.1 Discussion: Diversity, Equity, and Inclusion in Data Science (https://canvas.uw.edu/courses/1830727/discussion_topics/9915098) Due Oct 26, 11:59 PM
6.4 Discussion: Renaming Common Distributions (https://canvas.uw.edu/courses/1830727/discussion_topics/9915097) Due Nov 2, 11:59 PM
7.2.1 Discussion: Visual Introduction to Statistical Learning (https://canvas.uw.edu/courses/1830727/discussion_topics/9915096) Due Nov 9, 11:59 PM
9.9.1 Discussion: Data Science in Practice (https://canvas.uw.edu/courses/1830727/discussion_topics/9915095) Due Nov 23, 11:59 PM
"""


def parse_due_date(due_str, year='2025'):
    # Example: 'Oct 5, 11:59 PM'
    due_str = due_str.split(',')[0]
    if year not in due_str:
        due_str = due_str + f' {year}'
    return datetime.strptime(due_str, '%b %d %Y')

pattern = r'^(.*?)\s*\((https://[^\)]+)\)\s*Due\s*([A-Za-z]+\s+\d+,\s+\d{1,2}:\d{2}\s*[AP]M)'
rows = []
for line in text.strip().split('\n'):
    m = re.match(pattern, line)
    if m:
        title = m.group(1).strip()
        url = m.group(2).strip()
        due = m.group(3).strip()
        rows.append({'Title': title, 'URL': url, 'Due Date': due})

df = pd.DataFrame(rows)

with pd.ExcelWriter('discussions.xlsx', engine='xlsxwriter') as writer:
    df.to_excel(writer, index=False, sheet_name='Discussions', startrow=1, header=False)
    workbook  = writer.book
    worksheet = writer.sheets['Discussions']
    header = ['Title', 'URL', 'Due Date']
    for col_num, value in enumerate(header):
        worksheet.write(0, col_num, value)
    for row_num, row in enumerate(rows, start=1):
        # formula = f'=HYPERLINK("{row["URL"]}", "{row["Title"]}")'
        due_dt = parse_due_date(row['Due Date'])
        worksheet.write_url(row_num, 0, row['URL'], string=row['Title'])
        worksheet.write(row_num, 1, 'discussion')
        worksheet.write(row_num, 2, due_dt)
```
